```python
import numpy as np

def eliminacao_gauss(A, b):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    n = len(b)

    # Matriz aumentada
    M = np.column_stack((A, b))

    # Eliminação de Gauss
    for k in range(n - 1):
        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]

            for j in range(k, n + 1):
                M[i, j] = M[i, j] - m * M[k, j]

    # Substituição regressiva
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        soma = 0

        for j in range(i + 1, n):
            soma += M[i, j] * x[j]

        x[i] = (M[i, n] - soma) / M[i, i]

    return x


A = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b = [8, -11, -3]

x = eliminacao_gauss(A, b)

print("Solução pelo método de Gauss:")
print(x)

print("\nSolução pelo numpy:")
print(np.linalg.solve(A, b))
```

Os dois métodos apresentaram a mesma solução, \(x=2\), \(y=3\) e \(z=-1\). Portanto, o algoritmo de eliminação de Gauss implementado apresentou o mesmo resultado que a função numpy.linalg.solve

```python
import numpy as np

A = np.array([
    [4, -1, 0],
    [-1, 4, -1],
    [0, -1, 3]
], dtype=float)

b = np.array([15, 10, 10], dtype=float)

T = np.linalg.solve(A, b)

print("Temperaturas:")
print("T1 =", T[0])
print("T2 =", T[1])
print("T3 =", T[2])
```
